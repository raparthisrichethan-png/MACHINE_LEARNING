## DBSCAN (Density-Based)

- Two Parameters: 
    1) Epislon
    2) min point
        - core points
        - boundary points
        - Noise points
- Anomaly detection:
- Isolation Forest 
   > used for outlier detection
   > 

   - if path length is short , anomaly point
- if anomaly score = 1 (strong abnormal) , anomaly score = 0 (Normal Datapoint)
- S(x,n)=E|h(x)|/C(n) (Average path length of the data point)  C(n) - (avg path of trees)
- Algorithm Steps:
    ### For each tree
    1) Randomly select n sample
    2) Build tree using Random split
    3) compute path length for each datapoint.

    ### After building all trees
    1) compute avg path length
    2) normalize using C(n)
    3) compute anomaly score
    4) classify using contimination threshold
