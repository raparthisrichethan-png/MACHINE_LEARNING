```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('embedded_system_network_security_dataset.csv')
df.head()
```

```python
df.info()
```

```python
features=df.drop('label',axis=1)
features
```

```python
# step 2: Data Preprocessing

for col in features.columns:
    if features[col].dtype == 'bool':
        features[col]=features[col].astype(int)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_features=scaler.fit_transform(features)
```

```python
scaled_features_df=pd.DataFrame(scaled_features,columns=features.columns)
scaled_features_df.head()
```

```python
from sklearn.ensemble import IsolationForest
n_estimators=100
contamination = 0.01
sample_size= 256

model=IsolationForest(n_estimators=n_estimators,contamination=contamination,
max_samples=sample_size,random_state=42)
model.fit(features)

anomaly_labels = model.predict(features)
# -1 FOR anomalies, 1 for normal instances

scaled_features_df['anomaly']=anomaly_labels
scaled_features_df.head()

c=scaled_features_df['anomaly'].value_counts()
print(c)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 6))
normal = scaled_features_df[scaled_features_df['anomaly'] == 1]
anomaly = scaled_features_df[scaled_features_df['anomaly'] == -1]

# plot 1
axes[0].scatter(normal['packet_size'], normal['inter_arrival_time'], color='blue', label='Normal', alpha=0.5)
axes[0].scatter(anomaly['packet_size'], anomaly['inter_arrival_time'], color='red', label='Anomaly', alpha=0.5)
axes[0].set_title('Normal Data')
axes[0].set_xlabel('Packet Size')
axes[0].set_ylabel('inter_arrival_time')
axes[0].legend()
axes[0].grid(True,alpha=0.3)
# plot 2
axes[1].scatter(normal['src_port'],normal['dst_port'], color='blue', label='Normal', alpha=0.5)
axes[1].scatter(anomaly['src_port'], anomaly['dst_port'], color='red', label='Anomaly', alpha=0.5)
axes[1].set_title('Anomaly Data')
axes[1].set_xlabel('src_port')
axes[1].set_ylabel('dst_port')
axes[1].legend()
axes[1].grid(True,alpha=0.3)
plt.tight_layout()
plt.show()
```
